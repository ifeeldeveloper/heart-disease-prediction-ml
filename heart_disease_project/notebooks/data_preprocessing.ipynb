#### Heart Disease Data Analysis using Python
This project analyzed a heart disease dataset using Python and common data analysis libraries such as NumPy, Pandas, Matplotlib, and Seaborn. The data was cleaned by handling missing values, removing duplicates, correcting inconsistencies, and standardizing data types. Exploratory Data Analysis (EDA) and visualizations were then performed to understand participant demographics, lifestyle factors, health indicators, and their relationships with heart disease. The project demonstrates a complete data analysis workflow from raw data preparation to extracting meaningful insights.
